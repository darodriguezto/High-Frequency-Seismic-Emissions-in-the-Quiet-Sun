```python
import drms

client = drms.Client()

series_list = [
    "hmi.B_720s",
    "hmi.B_720s_dcon",
    "hmi.B_720s_dconS",
    "hmi.Bharp_720s",
    "hmi.Bharp_720s_nrt",
    "hmi.Bptr_720s_mod"
]

for series in series_list:
    print("=" * 100)
    print(f"SERIES: {series}")
    print("=" * 100)

    try:
        info = client.info(series)

        print("\nDESCRIPTION:")
        print(info.note)

        #print("\nPRIME KEYS:")
        #print(info.primekeys)

        #print("\nSEGMENTS:")
        #print(info.segments)

    except Exception as e:
        print(f"\nCould not retrieve information: {e}")

    print("\n")

```
