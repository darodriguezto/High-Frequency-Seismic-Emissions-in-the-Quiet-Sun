```python
from sunpy.net import Fido, attrs as a

# ------------------------------------------------------------
# 1. Define the observing time
# ------------------------------------------------------------

time = a.Time(
    "2018-08-25 09:12:00",
    "2018-08-26 09:12:00"
)

# ------------------------------------------------------------
# 2. Select the JSOC data series
# ------------------------------------------------------------

series = a.jsoc.Series("hmi.B_720s")

# ------------------------------------------------------------
# 3. Select only the magnetic-field magnitude
# ------------------------------------------------------------

segment = a.jsoc.Segment("field")

# JSOC registered email
email = a.jsoc.Notify("darodriguezto@unal.edu.co")

# ------------------------------------------------------------
# 4. Search for the data
# ------------------------------------------------------------

result = Fido.search(
    time,
    series,
    segment,
    email
)

print(result)

# ------------------------------------------------------------
# 5. Define the download directory
# ------------------------------------------------------------

path = "../data/2018_08_25/vector_field/{file}"

# ------------------------------------------------------------
# 6. Download the files
# ------------------------------------------------------------

downloaded_files = Fido.fetch(
    result,
    path=path
)

print(downloaded_files)
```

```python
## Pasar archivos fits a la extensión adecuada para el pipeline correspondiente ## 

from pathlib import Path
from astropy.io import fits

# Carpeta con los FITS descargados de JSOC
input_dir = Path("../data/2018_08_25/vector_field")

# Carpeta donde guardaremos los FITS compatibles con el JOB
output_dir = input_dir / "field_primary"
output_dir.mkdir(exist_ok=True)

# Buscar únicamente los archivos hmi.B_720s del segmento field
files = sorted(input_dir.glob("hmi.b*.field.fits"))

print(f"Archivos encontrados: {len(files)}")

for file in files:

    with fits.open(file) as hdul:

        # La imagen está en la extensión 1
        data = hdul[1].data
        header = hdul[1].header.copy()

        # Pasar la imagen al Primary HDU
        output_file = output_dir / file.name

        fits.PrimaryHDU(
            data=data,
            header=header
        ).writeto(output_file, overwrite=True)

    print(f"✓ {file.name}")

print(f"\nConversión terminada.")
print(f"Archivos guardados en:\n{output_dir.resolve()}")
```
