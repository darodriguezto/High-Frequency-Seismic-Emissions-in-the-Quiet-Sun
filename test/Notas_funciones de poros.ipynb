```python
# Extract pore centroids into an (N, 2) array, with one row per pore
# and columns corresponding to the x and y pixel coordinates.
# Reshape ensures a consistent (0, 2) shape when no pores are detected.
# The final matrix of distance shape is (N,M) N is the # of pores in the first frame and M the # of pores of the second one
import numpy as np
from scipy.spatial.distance import cdist

# Get selected regions from two reference frames
pores_0 = magnetograms_of_reference[0]["selected_regions"]
pores_480 = magnetograms_of_reference[1]["selected_regions"]

# Extract centroid coordinates
centroids_0 = np.array([
    [region["x_center"], region["y_center"]]
    for region in pores_0
]).reshape(-1, 2)

centroids_480 = np.array([
    [region["x_center"], region["y_center"]]
    for region in pores_480
]).reshape(-1, 2)

# Compute pairwise Euclidean distances
distance_matrix = cdist(
    centroids_0,
    centroids_480,
    metric="euclidean"
)

print("Distance matrix shape:", distance_matrix.shape)
print(distance_matrix)
```
