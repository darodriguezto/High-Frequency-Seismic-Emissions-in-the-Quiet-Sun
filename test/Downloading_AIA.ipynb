```python
import astropy.units as u
from sunpy.net import Fido, attrs as a
import sunpy.map
import matplotlib.pyplot as plt

# --------------------------------------------------
# Observation parameters
# --------------------------------------------------

start_time = "2018-08-25 09:07:48"  # Replace with MiHI start time
end_time   = "2018-08-25 09:23:50"  # Replace with MiHI end time

wavelength = 171 * u.AA

# Replace with your registered JSOC email
jsoc_email = "darodriguezto@unal.edu.co"

# --------------------------------------------------
# Search AIA observations
# --------------------------------------------------

result = Fido.search(
    a.Time(start_time, end_time),
    a.jsoc.Series("aia.lev1_euv_12s"),
    a.Wavelength(wavelength),
    a.jsoc.Segment("image"),
    a.jsoc.Notify(jsoc_email)
)

print(result)

# --------------------------------------------------
# Download FITS files
# --------------------------------------------------

files = Fido.fetch(
    result,
    path="../data/2018_08_25/AIA/{file}",
    max_conn=4
)

print(f"Downloaded files: {len(files)}")
```
